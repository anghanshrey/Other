```python
data.replace([np.inf,-np.inf,"inf","-inf"],np.nan,inplace=True)
```

```python
data['JoiningDate'] = pd.to_datetime(data['JoiningDate'], errors='coerce')
```

```python
data.rename({
    "Name" : "name"
}, inplace=True)
```

```python
data.dtypes
```

```python
data.isna().sum()
```

```python
mean_value = ["Age", "Bonus", "PerformanceRating", "Experience"]

for col in mean_value:
    data[col] = data[col].fillna(data[col].mean())

data.isna().sum()
```

```python
mediun = ["Salary", "ProjectsCompleted"]

for col in mediun:
    data[col] = data[col].fillna(data[col].median())
```

```python
data['JoiningDate'] = data['JoiningDate'].fillna(pd.Timestamp.now().normalize())

data.isna().sum()
```

```python
mode = ["Department", "City"]

for col in mode:
    data[col] = data[col].fillna(data[col].mode()[0])
```

```python
import matplotlib.pyplot as plt
```

```python
department_sales = data.groupby("Department")['Salary'].sum()

print(department_sales)
```

```python
plt.figure(figsize=(10,6))

bars = plt.bar(
    department_sales.index,
    department_sales.values,
    color = "orange",
    label = "sales"
)

for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{height:,.0f}",
        ha="center",
        va="bottom"
    )

plt.title("Departent & sales data")
plt.xlabel("Department")
plt.ylabel("Sales")

plt.xlim(-0.5, len(department_sales) -0.5)
plt.ylim(0,200000000)
plt.yticks(
    [0, 25000000, 50000000, 75000000,
     100000000, 125000000, 150000000, 175000000],
    ["0", "2.5 Cr", "5 Cr", "7.5 Cr",
     "10 Cr", "12.5 Cr", "15 Cr", "17.5 Cr"]
)
plt.legend(loc="upper left", bbox_to_anchor=(0.9, 1.2))
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.xticks(rotation=320)

plt.tight_layout()
plt.show()
```
