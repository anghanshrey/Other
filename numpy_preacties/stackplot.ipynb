```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.DataFrame({
    "Month": ["January", "February", "March", "April", "May", "June"],
    "Google Search": [1200, 1500, 1800, 2100, 2400, 2800],
    "Social Media": [800, 950, 1100, 1300, 1500, 1750],
    "Direct Visitors": [600, 700, 850, 1000, 1150, 1300]
})

plt.figure(figsize=(10, 6))


plt.stackplot(
    data["Month"],
    data["Google Search"],
    data["Social Media"],
    data["Direct Visitors"],
    colors=["green", "black", "skyblue"],
    labels=["Google Search", "Social Media", "Direct Visitors"]
)

plt.title("Traffic Sources Chart")
plt.legend(loc="upper left")
plt.xlabel("Month")
plt.ylabel("Visitors")

plt.show()
```
