```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
print("Create a DataFrame with a 'Color' column having repeated values.")

df = pd.DataFrame({
    "Name" : ["Rahul", "Amit", "Zeel", "Riya", "Priya", "Pooja", "Neha", "Karan"],
    "Color" : ["Red", "Blue", "Red", "Green", "Red", "Blue", "Green", "Red"]
})

print(df)
```

```python
print(df.shape)
print(df.columns)
print(df.index)
print(df.dtypes)
print(df.values)
print(df.size)
print(df.ndim)
```

```python
df["Color"] = df["Color"].astype("category")

print(df)

print(df['Color'].dtype)
```

```python
unique_Count = df["Color"].unique()

print(unique_Count)
```

```python
count_color = df["Color"].value_counts()

print(count_color)
```

```python
percentage = df["Color"].value_counts(normalize=True) * 100

print(percentage)
```

```python
df["Size"] = ["Medium", "Small", "Large", "Medium", "Small", "Large", "Small", "Large"]

print(df)
```

```python
size_Order = ["Medium", "Small", "Large"]

df["Size"] = pd.Categorical(
    df["Size"],
    categories= size_Order
)

df_sorted = df.sort_values("Size")

print(df_sorted)
print(df)
```

```python
df["Size"] = df["Size"].astype("category")
```

```python
df.dtypes
```

```python
df["Size"] = df["Size"].cat.rename_categories({
    "Small" : "S",
    "Medium" : "M",
    "Large" : "L"
})

print(df.dtypes)
print(df)
```

```python
df["Size"] = df["Size"].astype(str)

print(df.dtypes)
```
