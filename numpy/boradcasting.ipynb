```python
import numpy as np
```

##### Basic Filtering

```python
ages = np.array([25, 30, 18, 44, 21, 17, 21])

adults = ages[ages >= 18]

print(adults)
```

```python
adults_age = ages[(ages > 20) & (ages < 35)]

print(adults_age)
```

```python
names = np.array(["Rahul" , "Alice", "Bob", "Carlice" , "Akash"])

selected_names = names[np.char.startswith(names, "A") | np.char.startswith(names, "R")]

print(selected_names)
```

#### Basic Masking Selection

```python
temp = np.array([28, 32 , 35, 30, 22])

is_hot = temp > 30

hot_temp = temp[is_hot]

print(temp)
print(is_hot)
print(hot_temp)
```

#### Masing for Modification

```python
temp[is_hot] = 31

print(temp)
```

##### np.where

```python

scores = np.array([68 , 81 , 78 , 90 , 55 , 78])

# If score >= 75 -> Pass , otherwise Fail

grades = np.where(scores >= 75 , "Pass" , "")

print(scores)
print(grades)
```

#### Reshaping a 1D to 2D

```python
num_1d = np.array([1, 2, 3, 4])

num_2d = num_1d.reshape(1,4)

print(num_1d)

print(num_2d)
```

##### Broandcasting a 1D Array to a 2D Array

```python
sales = np.array([[10, 20], [15,25], [12, 10]])

add_value = np.array([2, 3, 1])

print(sales)

print(add_value)
```

```python
bonues_col = add_value.reshape(3,1)

sales_with_bounes = sales + bonues_col

print(sales_with_bounes)
```

```python
print(bonues_col)

sales_flatten = sales_with_bounes.flatten()

print(sales_flatten)
```
