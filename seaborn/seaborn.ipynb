```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = sns.load_dataset("healthexp")

sns.lineplot(
    x="Year",
    y="Life_Expectancy",
    data = data,
    hue="Country"
)

plt.title("Life Expectancy Over the Years")
plt.xlabel("Year")
plt.ylabel("Life Expectancy")
plt.show()
```

```python
data = sns.load_dataset("healthexp")

sns.barplot(
    x="Country",
    y="Spending_USD",
    data=data,
    hue="Year"
)

plt.title("Healthcare Spending by Country")
plt.xlabel("Country")
plt.ylabel("Spending (USD)")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.scatterplot(
    x="Spending_USD",
    y="Life_Expectancy",
    data=data,
    hue="Country"
)

plt.title("Healthcare Spending vs Life Expectancy")
plt.xlabel("Spending (USD)")
plt.ylabel("Life Expectancy")
plt.show()
```

```python
sns.histplot(
    x="Life_Expectancy",
    data=data,
    bins=20,
    kde=True,
)

plt.title("Distribution of Life Expectancy")
plt.xlabel("Life Expectancy")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.boxplot(
    x="Country",
    y="Spending_USD",
    data=data,
    hue="Country"
)

plt.title("Healthcare Spending Distribution by Country")
plt.xlabel("Country")
plt.ylabel("Spending (USD)")
plt.xticks(rotation=45)
plt.yticks(
    [0, 2000, 4000, 6000, 8000, 10000, 12000],
    ["0 USD", "2000 USD", "4000 USD", "6000 USD", "8000 USD", "10000 USD", "12000 USD"]
)

plt.show()
```

```python
correlation = data[["Year", "Spending_USD", "Life_Expectancy"]].corr()

print(correlation)

plt.figure(figsize=(6, 4))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linecolor="skyblue",
    linewidths=5,
    square=True
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.countplot(
    x="Country",
    data=data,
    hue="Country"
)

plt.title("Number of Records by Country")
plt.xlabel("Country")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.violinplot(
    x="Country",
    y="Life_Expectancy",
    data=data
)

plt.title("Life Expectancy Distribution by Country")
plt.xlabel("Country")
plt.ylabel("Life Expectancy")
plt.xticks(rotation=45)
plt.show()
```
